```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\ayush\Documents\customer_behavior\customer_shopping_behavior.csv")
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(
    lambda x: x.fillna(x.median())                                                         #filling missing values or null values
)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()

df.columns = df.columns.str.replace(' ', '_')                                      #renaming all the coloumns names as lower case for easy use

df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})                               
```

```python
#creating age group 
labels = ['young_adult', 'adult', 'middle_aged', 'senior']
df['age_groups'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_groups'
```

```python
df = df.drop('promo_code_used',axis=1)      #bcz same as discound applied every person who applied discount had applied promo code
```

```python
#giving a numeric value to te coloumn frequency_of_purchases
df['frequency_of_purchases'].unique()
```

```python
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}
```

```python
df['purchases_frequency']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchases_frequency']].head()
```

```python
print(df['purchases_frequency'].unique())
```

```python
df['purchases_frequency'] = df['purchases_frequency'].fillna(0)
```

```python
print(df['purchases_frequency'].isnull().sum())
```

```python
df.to_csv(r"C:\Users\Ayush\Documents\customer_behavior\customer_cleaned.csv", index=False)
```
